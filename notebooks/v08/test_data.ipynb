## GMI

```python
from pansat.time import TimeRange
from pansat.geometry import LonLatRect
from pansat.catalog import Index
from pansat.products.satellite.gpm import l1c_r_gpm_gmi_b

lat_min = 15
lat_max = 25
lon_min = -95
lon_max = -85
roi = LonLatRect(lon_min, lat_min, lon_max, lat_max)

start_date = np.datetime64("2024-10-08")
end_date = np.datetime64("2024-10-09")
recs = l1c_r_gpm_gmi_b.get(TimeRange(start_date, end_date))
index = Index.index(l1c_r_gpm_gmi_b, recs)
```

```python
granules = index.find(roi=roi)
print(f"Found {len(granules)} overpasses.")
```

```python
from copy import copy
f, ax = plt.subplots(1, 1, figsize=(7, 5))

granule = copy(granules[0])
offset = 40
scan_start, scan_end = granule.primary_index_range
granule.primary_index_range = (scan_start + offset, scan_end + offset)

obs = granule.open()
lons = obs.longitude_s1.data
lats = obs.latitude_s1.data

m = ax.pcolormesh(lons, lats, obs.tbs_s1[..., -1])
```

```python
from gprof_nn.data.preprocessor import run_preprocessor
from gprof_nn import sensors

l1c_file = Path("1C-R.GPM.GMI.XCAL2016-C.20241008-S051110-E064423.060255.V07B.HDF5")
pp_file = l1c_file.with_suffix(".nrt.pp")
run_preprocessor(
    l1c_file,
    sensors.GMI,
    output_file=pp_file,
    settings={
        "prodtype": "NRT",
        "prepdir": "/qdata1/pbrown/gpm/modelprep/GANALV7/"
    }
)
```
