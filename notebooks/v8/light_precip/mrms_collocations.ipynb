```python
from gprof_nn.data.utils import merge_precipitation
from tqdm import tqdm

path_merged = Path("/gdata1/simon/gprof_v8/models/gmi/gprof_nn_light/results_merged")
path_light = Path("/gdata1/simon/gprof_v8/models/gmi/gprof_nn_light/results_light")
results = sorted(list(Path("/gdata1/simon/gprof_v8/models/gmi/gprof_nn_light/results").glob("**/*.nc")))
path_merged.mkdir(exist_ok=True)
path_light.mkdir(exist_ok=True)

for path in tqdm(results):
    res = xr.load_dataset(path)
    res_light = res[["latitude", "longitude", "light_precip"]].rename(light_precip="surface_precip")
    res_light.to_netcdf(path_light / path.name)
    precip_merged = merge_precipitation(res.surface_precip, res.light_precip)
    res_merged = xr.Dataset({
        "surface_precip": precip_merged,
        "latitude": res.latitude,
        "longitude": res.longitude
    })
    res_merged.to_netcdf(path_merged / path.name)
```

```python
from gprof_nn.validation import Evaluator

ref_path = Path("/xdata/simon/gpm_collocations_west/gmi/on_swath/")
res_path_cmb = Path("/gdata1/simon/gprof_v8/models/gmi/gprof_nn_light/results")
res_path_light = Path("/gdata1/simon/gprof_v8/models/gmi/gprof_nn_light/results_light/")
res_path_merged = Path("/gdata1/simon/gprof_v8/models/gmi/gprof_nn_light/results_merged/")

evaluator = Evaluator(
    ref_path,
    {
        "CMB": res_path_cmb,
        "CS": res_path_light,
        "CMB + CS": res_path_merged
    }
)
    
```

```python
results_land, results_ocean = evaluator.evaluate()
```
